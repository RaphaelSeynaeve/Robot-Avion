```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from scipy.signal import convolve2d, windows

def gaussian_kernel(size=10, sigma=3):
    g = windows.gaussian(size, std=sigma)
    G = np.outer(g, g)
    return G / G.sum()

# Load and convert image to grayscale array
gray_lena = np.array(Image.open('Lena.png').convert('L'))

# Create Gaussian kernel and convolve
G = gaussian_kernel(10, 3)
gaussian_lena = convolve2d(gray_lena, G, mode='same')
plt.imshow(gaussian_lena, cmap='gray')
plt.show()
```

```python
plt.title('Smoothed Image')
plt.axis('off')
plt.imshow(gaussian_lena, cmap='gray')
plt.show()
plt.show()

# Display difference
plt.imshow(gray_lena - gaussian_lena, cmap='gray')
plt.title('Original Image - Smoothed Image')
plt.axis('off')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 2D Gaussian kernel using meshgrid ---
def gaussian_kernel(size=21, sigma=5):
    ax = np.linspace(-(size // 2), size // 2, size)
    xx, yy = np.mes_grid(ax, ax)
    G = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    return G / G.sum()

G = gaussian_kernel(size=21, sigma=5)

plt.imshow(G, cmap='gray')
plt.colorbar()
plt.title('2D Gaussian Kernel')
plt.axis('off'); plt.show()
```
